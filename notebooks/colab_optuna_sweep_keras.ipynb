# Hyperparameter Sweep (Optuna) - Keras CAE

**NOTE**: Running sequential TensorFlow Optuna studies within a single persistent notebook kernel will eventually cause memory accumulation (both GPU VRAM and System RAM) due to TensorFlow graph execution and dataset caching.

To prevent resource exhaustion and crashing, this notebook delegates the sweep to `scripts/sweep.py --model keras`. The runner spawns an isolated subprocess for each category with `TF_GPU_ALLOCATOR=cuda_malloc_async`, guaranteeing that the OS reclaims 100% of memory after each category finishes.

Results are incrementally persisted to `data/hyperparameters/keras_cae_best.json`.

```python
import os
import sys
from pathlib import Path

# Ensure we are in the project root (supports both local environment and Google Colab)
current_dir = Path.cwd()
while not (current_dir / "app").exists() and current_dir != current_dir.parent:
    current_dir = current_dir.parent
PROJECT_ROOT = current_dir
os.chdir(PROJECT_ROOT)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f"Project root resolved: {PROJECT_ROOT.resolve()}")
```

```python
import json

# Inspect saved hyperparameter configurations
results_path = PROJECT_ROOT / "data/hyperparameters/keras_cae_best.json"
if results_path.exists():
    results = json.loads(results_path.read_text(encoding="utf-8"))
    print(f"Completed categories ({len(results)}/15): {list(results.keys())}\n")
    print(json.dumps(results, indent=2))
else:
    print(f"No hyperparameter file found at {results_path}")
```
