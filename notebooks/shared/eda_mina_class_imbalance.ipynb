```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_ROOT = Path("../../data/raw/mvtec_ad")
categories = sorted([d.name for d in DATA_ROOT.iterdir() if d.is_dir() and (d / "test").exists()])
print(categories)
```

```python
records = []

for category in categories:
    test_dir = DATA_ROOT / category / "test"
    for subfolder in sorted(test_dir.iterdir()):
        if not subfolder.is_dir():
            continue
        n_images = len(list(subfolder.glob("*.png")))
        label = "good" if subfolder.name == "good" else "defective"
        records.append({"category": category, "subfolder": subfolder.name, "label": label, "count": n_images})

df = pd.DataFrame(records)
df.head(20)
```

```python
category_counts = df.groupby(["category", "label"])["count"].sum().unstack(fill_value=0)
category_counts["total"] = category_counts["good"] + category_counts["defective"]
category_counts["defective_ratio"] = category_counts["defective"] / category_counts["total"]
category_counts
```

```python
from scipy.stats import chi2_contingency

contingency_table = category_counts[["good", "defective"]]
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square statistic: {chi2:.3f}")
print(f"Degrees of freedom: {dof}")
print(f"p-value: {p_value:.2e}")
```

```python
import numpy as np
from PIL import Image

sample_mask_path = DATA_ROOT / "bottle" / "ground_truth" / "broken_large" / "000_mask.png"
mask = np.array(Image.open(sample_mask_path))

print("Shape:", mask.shape)
print("Unique values:", np.unique(mask))
print("Anomalous pixels:", (mask > 0).sum())
print("Total pixels:", mask.size)
```

```python
pixel_records = []

for category in categories:
    gt_dir = DATA_ROOT / category / "ground_truth"
    if not gt_dir.exists():
        continue
    for defect_type_dir in sorted(gt_dir.iterdir()):
        if not defect_type_dir.is_dir():
            continue
        for mask_path in sorted(defect_type_dir.glob("*_mask.png")):
            mask = np.array(Image.open(mask_path))
            anomalous = int((mask > 0).sum())
            normal = int(mask.size - anomalous)
            pixel_records.append({
                "category": category,
                "defect_type": defect_type_dir.name,
                "anomalous_pixels": anomalous,
                "normal_pixels": normal
            })

pixel_df = pd.DataFrame(pixel_records)
pixel_df.head(10)
```

```python
category_pixel_totals = pixel_df.groupby("category")[["anomalous_pixels", "normal_pixels"]].sum()
category_pixel_totals["total_pixels"] = category_pixel_totals["anomalous_pixels"] + category_pixel_totals["normal_pixels"]
category_pixel_totals["anomalous_ratio"] = category_pixel_totals["anomalous_pixels"] / category_pixel_totals["total_pixels"]
category_pixel_totals.sort_values("anomalous_ratio")
```

### Key Insight: Image-Level vs. Pixel-Level Imbalance Diverge

Image-level and pixel-level imbalance tell very different stories. A category like
**"screw"** might have plenty of defective *images*, but each defect covers only a
tiny fraction of the image (0.34% anomalous pixels). In contrast, **"metal_nut"**
defects are much larger and more pervasive within the image (14.5% anomalous pixels)
— a **~40x difference** in severity, despite both categories having similarly
proportioned counts of defective vs. good *images*.
###### Short Insight: image-level and pixel-level imbalance diverge sharply (screw ~0.3% vs metal_nut ~14.5%)

Chi-Square test cell on the pixel-level table:

```python
pixel_contingency_table = category_pixel_totals[["anomalous_pixels", "normal_pixels"]]
chi2_px, p_value_px, dof_px, expected_px = chi2_contingency(pixel_contingency_table)

print(f"Chi-square statistic: {chi2_px:.3f}")
print(f"Degrees of freedom: {dof_px}")
print(f"p-value: {p_value_px:.2e}")
```

with millions of pixels, chi-square tests become extremely sensitive — even tiny, practically meaningless differences would register as "significant" at this sample size. The p-value being essentially zero is expected and somewhat less informative here than the effect size itself — the actual 40x range in anomalous_ratio (0.34% to 14.5%) is the number that matters for the business argument, not the p-value alone. 
more explanation:
look at the category_pixel_totals table, sorted by anomalous_ratio
screw — lowest — anomalous_ratio = 0.003355 → 0.34%
metal_nut — highest — anomalous_ratio = 0.144888 → 14.5%
0.144888​/0.003355​≈43.2
So metal_nut's anomalous pixel ratio is about 43 times higher than screw's — that's where the "~40x" figure comes from. It's the ratio between the smallest and largest values in that anomalous_ratio column across all 15 categories.

Build the boxplot

```python
fig, ax = plt.subplots(figsize=(12, 6))
pixel_df.boxplot(column="anomalous_ratio", by="category", ax=ax, rot=45)
ax.set_ylabel("Anomalous pixel ratio (per image)")
ax.set_title("Anomaly Size Distribution by Category")
plt.suptitle("")
plt.tight_layout()
plt.show()
```

```python

fig, ax = plt.subplots(figsize=(14, 7))

# Order categories by median, so the chart tells a clearer story left-to-right
order = pixel_df.groupby("category")["anomalous_ratio"].median().sort_values().index
data_ordered = [pixel_df[pixel_df["category"] == cat]["anomalous_ratio"].values for cat in order]

order = list(pixel_df.groupby("category")["anomalous_ratio"].median().sort_values().index)
data_ordered = [pixel_df[pixel_df["category"] == cat]["anomalous_ratio"].values for cat in order]


bp = ax.boxplot(data_ordered, tick_labels=order, patch_artist=True, showmeans=False)

# Color the boxes
for box in bp['boxes']:
    box.set_facecolor("#4C72B0")
    box.set_alpha(0.6)

# Log scale fixes the "everything squished near 0" problem
ax.set_yscale("log")
ax.set_ylabel("Anomalous pixel ratio (per image, log scale)")
ax.set_xlabel("Category")
ax.set_title("Anomaly Size Distribution by Category")

# Add median value labels above each box
medians = [d.mean() for d in data_ordered]  # or use np.median(d)
import numpy as np
for i, d in enumerate(data_ordered):
    med = np.median(d)
    ax.text(i + 1, med * 1.3, f"{med:.1%}", ha="center", fontsize=8, color="darkred")

plt.xticks(rotation=45, ha="right")
ax.grid(True, which="both", ls="--", alpha=0.4)
plt.tight_layout()
plt.savefig("figure_7a_anomaly_size_distribution.png", dpi=150, bbox_inches="tight")
plt.show()
```
